```python

SIZE = 13

n = int(input("Enter number of keys: "))
keys = list(map(int, input("Enter the keys: ").split()))

# Separate Chaining
chaining = [[] for _ in range(SIZE)]
for key in keys:
    chaining[key % SIZE].append(key)

# Linear Probing
linear = [None] * SIZE
for key in keys:
    index = key % SIZE
    while linear[index] is not None:
        index = (index + 1) % SIZE
    linear[index] = key

# Quadratic Probing
quadratic = [None] * SIZE
for key in keys:
    h = key % SIZE
    for i in range(SIZE):
        index = (h + i * i) % SIZE
        if quadratic[index] is None:
            quadratic[index] = key
            break

# Double Hashing
double = [None] * SIZE
for key in keys:
    h1 = key % SIZE
    h2 = 7 - (key % 7)

    for i in range(SIZE):
        index = (h1 + i * h2) % SIZE
        if double[index] is None:
            double[index] = key
            break

# Display all hash tables
print("\nHASH TABLE")
print("-" * 80)
print(f"{'Index':<8}{'Chaining':<20}{'Linear Probing':<18}"
      f"{'Quadratic Probing':<20}{'Double Hashing'}")
print("-" * 80)

for i in range(SIZE):
    print(f"{i:<8}{str(chaining[i]):<20}{str(linear[i]):<18}"
          f"{str(quadratic[i]):<20}{str(double[i])}")
```
